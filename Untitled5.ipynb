```python
from graphviz import Digraph
from IPython.display import display

def create_workflow_diagram():
    # Directed Graph
    dot = Digraph('OptimizedWorkflow', format='png')
    dot.attr(rankdir='TB', compound='true')
    dot.attr(fontname='Arial')

    # Style mặc định cho các node
    dot.attr('node', shape='box', style='filled, rounded', fontname='Arial', margin='0.2')

    # --- TẦNG ĐẦU VÀO ---
    with dot.subgraph(name='cluster_input') as c:
        c.attr(label='Tầng Đầu Vào & Tự Động Hóa', style='dashed', color='gray')
        c.node('M1', 'Module 1: Ứng dụng tạo mã QR\n(Người thao tác: Sale)\n• Tạo Info Code\n• Ghi nhận thông tin KH, UID, Số tiền',
               fillcolor='#e1f5fe', color='#03a9f4', penwidth='2')
        c.node('M2', 'Module 2: API Ngân hàng / Webhook\n(Hệ thống tự động - Real-time)\n• Nhận biến động số dư tức thời\n• Tự động đối chiếu qua Info Code',
               fillcolor='#e1f5fe', color='#03a9f4', penwidth='2')

    # --- TẦNG LƯU TRỮ ---
    with dot.subgraph(name='cluster_storage') as c:
        c.attr(label='Tầng Dữ Liệu Tập Trung', style='dashed', color='gray')
        c.node('DW', 'Unified Data Warehouse\n(Gộp DW1 và DW2)\n• Lưu thông tin KH & Giao dịch\n• Lưu dữ liệu phễu CRM',
               shape='cylinder', fillcolor='#ffebee', color='#ef5350', penwidth='2')

    # --- HỆ THỐNG CRM ---
    with dot.subgraph(name='cluster_crm') as c:
        c.attr(label='Hệ thống CRM', style='dashed', color='gray')
        c.node('M3', 'Module 3: Tự động hóa CRM\n(Hệ thống tự động)\n• Gọi API tự động tạo đơn hàng\n• Tự động kích hoạt khóa học',
               fillcolor='#e8f5e9', color='#4caf50', penwidth='2')
        c.node('CRM_DB', 'Hệ Thống CRM Core\n(Dữ liệu trạng thái Học viên)',
               shape='cylinder', fillcolor='#e8f5e9', color='#4caf50', penwidth='2')

    # --- GIÁM SÁT & FALLBACK ---
    with dot.subgraph(name='cluster_monitor') as c:
        c.attr(label='Giám Sát & Xử Lý Sự Cố', style='dashed', color='gray')
        c.node('DB1', 'Dashboard 1: Kiểm soát vận hành\n• Theo dõi trạng thái đơn hàng\n• Phân quyền, bảo mật cao',
               fillcolor='#fff3e0', color='#ff9800', penwidth='2')
        c.node('Fallback', 'Quy trình Fallback (Thủ công)\n• Upload ảnh bill khi KH chuyển sai Info Code\n• Duyệt thủ công để kích hoạt',
               fillcolor='#fff3e0', color='#ff9800', penwidth='2')

    # --- TẦNG ĐẦU RA ---
    with dot.subgraph(name='cluster_output') as c:
        c.attr(label='Tầng Đầu Ra Tự Động', style='dashed', color='gray')
        c.node('EX', 'File Excel SM Hanoi và HCM Revenue', fillcolor='#f3e5f5', color='#ab47bc', penwidth='2')
        c.node('INV', 'Luồng Xuất Hóa Đơn Hàng Ngày', fillcolor='#f3e5f5', color='#ab47bc', penwidth='2')
        c.node('RP', 'Báo Cáo Hàng Ngày Sale Leader', fillcolor='#f3e5f5', color='#ab47bc', penwidth='2')

        c.node('INV1', 'Hóa đơn khách hàng', shape='ellipse', fillcolor='#f3e5f5', color='#ab47bc')
        c.node('INV2', 'Hóa đơn sản phẩm', shape='ellipse', fillcolor='#f3e5f5', color='#ab47bc')
        c.node('INV3', 'Hóa đơn đơn hàng', shape='ellipse', fillcolor='#f3e5f5', color='#ab47bc')

    # --- KẾT NỐI ---
    dot.edge('M1', 'DW', label=' Đẩy thông tin giao dịch')
    dot.edge('M2', 'DW', label=' Bắn dữ liệu khớp Info Code')

    dot.edge('DW', 'M3', label=' Xác nhận đã thanh toán')
    dot.edge('M3', 'CRM_DB', label=' Trigger API')
    dot.edge('CRM_DB', 'DW', label=' Đồng bộ phễu CRM')

    dot.edge('DW', 'DB1', label=' Hiển thị trạng thái')
    dot.edge('Fallback', 'DB1', label=' Xử lý ngoại lệ')
    dot.edge('DB1', 'DW', label=' Cập nhật thủ công')

    dot.edge('DW', 'EX')
    dot.edge('DW', 'INV')
    dot.edge('DW', 'RP')

    dot.edge('INV', 'INV1')
    dot.edge('INV', 'INV2')
    dot.edge('INV', 'INV3')

  
    dot.render('architecture_diagram', format='png', cleanup=True)

   
    return dot

# Gọi hàm và hiển thị
diagram = create_workflow_diagram()
display(diagram)
```
